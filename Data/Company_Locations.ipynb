# Fix loactions for mining companies

## Setup

```python
import pandas as pd
import re
```

## Make dataframe

```python
# -----------------------------
# 1. Data (zoals uit je tabel)
# -----------------------------
data = {
    "ticker": [
        "ARG.TO","AYA.TO","CG.TO","CGC.V","CKG.V","DMX.V","DV.V",
        "EDG.V","ELO.TO","EQX","ESK.V","FAS.V","FDY.TO","FINX.V",
        "FNM.V","FOM.TO","FURY.TO","FVL.TO","FWZ.V"
    ],
    "locations": [
        "Chile",
        "Morocco",
        "USA, Canada, Turkey",
        "Canda",
        "Mexico, USA",
        "Sweden",
        "Canada",
        "USA (Alaska), Canada",
        "Bolivia, Peru, Canada (Quebec)",
        "Canada (Ontario), USA, Canada, Nicaragua, Brazil",
        "Canada",
        "Peru",
        "USA",
        "Finland",
        "Sweden, Finland",
        "Canada",
        "Canada (east and west)",
        "USA (Colorado)",
        "Canada"
    ]
}
```

## Clean data

```python
df = pd.DataFrame(data)


def normalize_locations(loc_str):
    # alles splitten op komma
    locations = [l.strip() for l in loc_str.split(",")]

    cleaned = []
    for loc in locations:
        # verwijder alles tussen haakjes
        loc = re.sub(r"\(.*?\)", "", loc).strip()

        # typfouten & varianten normaliseren
        if loc.lower().startswith("can"):
            loc = "Canada"
        elif loc.lower().startswith("usa"):
            loc = "USA"

        cleaned.append(loc)

    # unieke waarden behouden
    return sorted(set(cleaned))
```

## Create dictionary and save as csv

```python
# --------------------------------
# 3. Dictionary maken
# --------------------------------
ticker_dict = {
    row["ticker"]: normalize_locations(row["locations"])
    for _, row in df.iterrows()
}

# --------------------------------
# 4. Opslaan als CSV
# --------------------------------
output_df = pd.DataFrame([
    {"ticker": k, "locations": "; ".join(v)}
    for k, v in ticker_dict.items()
])

output_df.to_csv("tickers_mining_locations.csv", index=False)

print("CSV opgeslagen als: tickers_mining_locations.csv")
print("\nVoorbeeld dictionary:")
print(ticker_dict)
```
